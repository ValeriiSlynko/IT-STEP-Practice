# Завдання 3
Виведіть дані про купівлі телевізорів та телефонів

```python
# дістаємо дані про купівлю телевізорів через створену маску "mask_television"
mask_television = df['item'] == 'Television'
# дістаємо дані про купівлю телефонів через створену маску "mask_call_phone"
mask_call_phone = df['item'] == 'Cell Phone'
# обєднуємо дві маски в одну
mask = mask_television | mask_call_phone
# виводимо рядки з умовами
df[mask]
```

# Завдання 6
Перевірте чи правильно розраховано загальну вартість покупки. Виведіть ті рядки де є помилка. Виправте дані в таблиці

```python
# для початку створюємо стовпчик з розрахованими покупками для перевірки
df['check_amt'] = df['units'] * df['unit_price']
df['check_amt']
```

```python
# виводимо рядки з помилками через маску для рахунку відклонінь
mask_error = df['check_amt'] != df['sale_amt']
# покаже True/False
mask_error
```

```python
# показуємо самі рядки з помилками
df[mask_error]
```

```python
# виправляємо дані з таблиці
# осн. стовпчик sale_amt перезаписуємо правильними значеннями із df['check_amt']
df['sale_amt'] = df['check_amt']
df
```
